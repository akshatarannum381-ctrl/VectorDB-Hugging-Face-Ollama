1. What is a Vector Database (VectorDB) and how is it different from
traditional databases?

- A Vector Database (VectorDB) is a specialized database designed to store, index, and search vector embeddings—numerical representations of data such as text, images, audio, or video. These embeddings are typically generated by machine learning models and capture the semantic meaning of the data.

| Feature                        | Vector Database                               | Traditional Database              |
| ------------------------------ | --------------------------------------------- | --------------------------------- |
| **Data Type**                  | High-dimensional vectors (embeddings)         | Structured / semi-structured data |
| **Search Type**                | Similarity search (semantic)                  | Exact match, keyword-based        |
| **Use Case**                   | AI, NLP, recommendation systems               | Business apps, transactions       |
| **Indexing**                   | ANN (Approximate Nearest Neighbor) algorithms | B-trees, hash indexes             |
| **Query Example**              | “Find similar documents”                      | “Find record with ID = 101”       |
| **Handling Unstructured Data** | Very efficient                                | Limited / indirect                |


2.  Explain the various types of VectorDBs available and describe their
suitability for different use cases.

- Vector databases aren’t all built the same they differ in architecture, storage model, and optimization goals, which makes each type better suited for specific use cases. A useful way to understand them is by grouping them into four major categories.

a. In-Memory Vector Databases

These databases store vectors entirely in RAM, enabling extremely fast similarity searches.

Characteristics
Ultra-low latency
Limited by available memory
Often used for real-time systems

Examples
FAISS (can run in-memory)
Redis

Best suited for

Real-time recommendation systems
Autocomplete and search suggestions
Online chatbots requiring instant retrieval

b. Disk-Based (Persistent) Vector Databases

These store vectors on disk or SSD, allowing them to scale to massive datasets.

Characteristics
High scalability
Slightly higher latency than in-memory
Built for durability and persistence

Examples
Milvus
Weaviate

Best suited for
Enterprise AI systems
Large-scale document search (millions/billions of vectors)
Data lakes with embeddings

c. Hybrid Vector Databases (Vector + Traditional Search)

These combine vector similarity search + keyword/full-text search.

Characteristics
Supports structured filters + semantic search
More flexible querying
Often integrates with existing DB systems

Examples
Elasticsearch
PostgreSQL

Best suited for
E-commerce search (filters + similarity)
Enterprise knowledge bases
Legal and research databases

3. Why is Chroma DB considered important in the context of AI/ML projects?
Describe its key features.

- Chroma (often called Chroma DB) is a lightweight, developer-friendly vector database designed specifically for AI/ML applications—especially those using LLMs.

At a practical level, Chroma solves a key problem in modern AI systems:

How do you store, search, and retrieve meaningful context efficiently from unstructured data?

Instead of relying on keyword matching, Chroma stores embeddings and enables semantic search, which is critical for systems like chatbots, recommendation engines, and Retrieval-Augmented Generation (RAG).

Key Features of Chroma DB

1. Embedding Storage & Management
2. Semantic (Similarity) Search
3. Persistent Storage
4. Metadata Filtering
5. Easy API & Python Integration
6. Collection-Based Organization

6.  Install and set up Chroma DB, and insert sample vector data for semantic
search.

- Here’s a clean, practical walkthrough to install Chroma, set it up, and insert sample vector data for semantic search.

1. Install and set up Chroma DB, and insert sample vector data for semantic
search.
pip install chromadb

2. Initialize Chroma DB

Create a simple Python script:
import chromadb


client = chromadb.Client()


collection = client.create_collection(name="my_collection")

3. Add Sample Data (with Embeddings)

from sentence_transformers import SentenceTransformer


model = SentenceTransformer('all-MiniLM-L6-v2')


documents = [
    "Artificial intelligence is transforming healthcare.",
    "Machine learning helps in data analysis.",
    "Cricket is the most popular sport in India.",
    "Deep learning is a subset of AI."
]


embeddings = model.encode(documents).tolist()


collection.add(
    documents=documents,
    embeddings=embeddings,
    ids=["1", "2", "3", "4"]
)

4. Perform Semantic Search

query = "How is AI used in medicine?"

query_embedding = model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=2
)

print(results)

5. Add Metadata (Optional but Powerful)

collection.add(
    documents=["Football is popular worldwide."],
    embeddings=model.encode(["Football is popular worldwide."]).tolist(),
    metadatas=[{"category": "sports"}],
    ids=["5"]
)

6. Persistent Storage (Save Data)

client = chromadb.Client(
    chromadb.config.Settings(
        persist_directory="./chroma_db"
    )
)

client.persist()

9. Implement a basic RAG (Retrieval-Augmented Generation) system using
Ollama with Llama3.

- Below is a minimal, working RAG (Retrieval-Augmented Generation) setup using a local LLM via Ollama with Llama 3, and a vector store using Chroma.

1) Install & start Ollama

curl -fsSL https://ollama.com/install.sh | sh


ollama pull llama3

2) Install Python dependencies

pip install chromadb ollama sentence-transformers

3) Create a simple RAG script

import chromadb
from sentence_transformers import SentenceTransformer
import ollama


client = chromadb.Client()
collection = client.create_collection(name="docs")

embed_model = SentenceTransformer("all-MiniLM-L6-v2")

documents = [
    "Artificial intelligence is used in healthcare for diagnosis.",
    "Machine learning helps analyze large datasets.",
    "Llama3 is a powerful large language model for text generation.",
    "Cricket is the most popular sport in India."
]

embeddings = embed_model.encode(documents).tolist()

collection.add(
    documents=documents,
    embeddings=embeddings,
    ids=[str(i) for i in range(len(documents))]
)


query = "How is AI used in medicine?"

query_embedding = embed_model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=2
)

retrieved_docs = results["documents"][0]


context = "\n".join(retrieved_docs)

prompt = f"""
Answer the question using the context below.

Context:
{context}

Question:
{query}
"""


response = ollama.chat(
    model="llama3",
    messages=[{"role": "user", "content": prompt}]
)

print("Answer:\n", response["message"]["content"])
